```python
# Creating an empty list to store all students
# Each student will be a dictionary inside this list
students = []

print("Empty student list created!")
print(f"Current students: {students}")
print(f"Number of students: {len(students)}")
```

```python
def add_student(roll_no, name, marks):
    """Adding a new student to the system"""
    
    # Creating a dictionary for the new student
    new_student = {
        "roll_no": roll_no,
        "name": name,
        "marks": marks
    }
    
    # Add to our list
    students.append(new_student)
    print(f"Student added: {name} (Roll: {roll_no}, Marks: {marks})")
    return 

# Test the function
add_student(101, "Akshay", 88)
add_student(102, "Bindu", 94)
add_student(103, "Kavana", 78)
add_student(104, "Sanjana", 65)
```

```python
def add_student(roll_no, name, marks):
    """Adding a new student with duplicate checking"""
    
    # Check if roll number already exists
    for student in students:
        if student["roll_no"] == roll_no:
            print(f" Roll number {roll_no} already exists for {student['name']}!")
            print(f"   Student NOT added.")
            return False
    
    # If no duplicate, add the student
    new_student = {
        "roll_no": roll_no,
        "name": name,
        "marks": marks
    }
    students.append(new_student)
    print(f" Student added: {name} (Roll: {roll_no}, Marks: {marks})")
    return True

# Test it
students = [] 

add_student(101, "Akshay", 88)
add_student(102, "Bindu", 94)
add_student(103, "Kavana", 78)
add_student(104, "Sanjana", 65)
add_student(105, "Tejas", 77)

# Trying adding duplicate
add_student(101, "Duplicate", 99)  # This will be rejected

view_all_students()
```

```python
def search_student(roll_no):
    """Finding a student by roll number"""
    
    for student in students:
        if student["roll_no"] == roll_no:
            print(f"\n Student Found!")
            print(f"   Roll No: {student['roll_no']}")
            print(f"   Name: {student['name']}")
            print(f"   Marks: {student['marks']}")
            return student
    
    print(f"\n Student with roll number {roll_no} not found!")
    return None

# Test searches
search_student(102)  # Should find Bindu
search_student(110) # Should not find
```

```python
def update_marks(roll_no, new_marks):
    """Update marks for a specific student"""
    
    for student in students:
        if student["roll_no"] == roll_no:
            old_marks = student["marks"]
            student["marks"] = new_marks
            print(f"\n Marks Updated for {student['name']}")
            print(f"   Old Marks: {old_marks}")
            print(f"   New Marks: {new_marks}")
            return 
    
    print(f"\n Student with roll number {roll_no} not found!")
    return

# Test update
update_marks(102, 95) #update the marks of roll number 102
update_marks(110, 94) #roll number not found

```

```python
def delete_student(roll_no):
    """Remove a student from the list"""
    
    for i, student in enumerate(students):
        if student["roll_no"] == roll_no:
            removed = students.pop(i)
            print(f"\n Student removed: {removed['name']} (Roll: {removed['roll_no']})")
            return 
    
    print(f"\n Student with roll number {roll_no} not found!")
    return 

# Test delete
delete_student(105) #delete student
delete_student(110) #roll number not found
```
